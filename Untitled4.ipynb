```python
!pip install pooch -q
import xarray as xr
ds = xr.tutorial.open_dataset("air_temperature")
ds
```

```python
ds.air.isel(time=0).plot()
```

```python
air_c = ds.air - 273.15
air_c.sel(lat=40, lon=260, method="nearest").plot()
```

```python
monthly = air_c.groupby("time.month").mean()
monthly.sel(month=7).plot()
```
